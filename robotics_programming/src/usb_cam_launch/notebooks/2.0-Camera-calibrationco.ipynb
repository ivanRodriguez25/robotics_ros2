# Camera calibration  
Information about the camera used to project camera information into a 2D image plan.  
  
* **Extrinsic calibration**: external information about the sensor. Rotation and translation of the camera with respect to the base frame of the robot.  
* **Intrinsic calibration**: internal characteristics of the sensor. Settings that specify how the camera creates an image on its sensor.  

Goal: to map pixel coordinates to correpsondoing points in the real world. It involves computing intrinsic parameters, such as focal length, and optimal center, that relate pixel coordinates to real-world points.     

```$sudo apt-get install ros-humble-camera-calibration```

```$ros2 run camera_calibration cameracalibrator --size 3x3 --square 0.0036 --ros-args -r image:=camera/image_raw -p camera:=usb_camera```

1. ```--size```: number of inner corners on the chessboard pattern
2. ```--square```: square sixz eof the chessboar din meters
3. ```--image```: topic in which the image is published
4. ```--camera_name```: string name for the camera
